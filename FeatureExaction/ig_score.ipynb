# Integrated Gradients (IG) Attribution Analysis

To understand **which parts of the variant annotation prompt most strongly influence model predictions**, we perform a comprehensive analysis of **token-level Integrated Gradients (IG) scores** extracted from ablated models.

Using the provided `ig_score_example.jsonl` (output from `GetEmb.py --IGscore`), we:

1. **Map raw token-level IG scores** back to semantic annotation categories (e.g., `"af"`, `"ACMG"`, `"IUIS"`).
2. **Aggregate attributions per category** per sample, optionally using absolute values (`use_abs_ig=True`) to focus on magnitude rather than direction.
3. Conduct four types of analysis:
   - **Distribution plots**: overall and per-category IG score distributions
   - **Category comparison**: statistical testing (Kruskal-Wallis + Dunn’s post-hoc) to identify which annotation types receive significantly higher attribution
   - **Dimensionality reduction**: PCA/UMAP on full token-level IG vectors to reveal sample clustering patterns
   - **Category-aggregated embedding**: PCA/UMAP on category-summarized attribution vectors

This pipeline reveals **which knowledge sources the model relies on most** during IEI variant prioritization—and whether this reliance varies across patients or ablation conditions.

> **Note**: All visualizations are saved automatically to `plots/abs_ig/` (or `plots/raw_ig/` if `use_abs_ig=False`).

```python
from igscore import run_full_analysis

JSONL_PATH = "ig_score_example.jsonl"
OUTPUT_ROOT = "plots"
# Run analysis with absolute IG (recommended for magnitude-based interpretation)
run_full_analysis(
    jsonl_path=JSONL_PATH,
    output_dir=OUTPUT_ROOT,
    use_abs_ig=True,
    save=False,
)


# Optional: also run with raw IG (to see signed contributions)
# run_full_analysis(jsonl_path=JSONL_PATH, output_dir=OUTPUT_ROOT, use_abs_ig=False)
```
